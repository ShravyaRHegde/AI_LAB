```python
# Tic-Tac-Toe using Minimax

board = [" " for _ in range(9)]

HUMAN = "X"
COMPUTER = "O"


def display_board():
    print()
    print(board[0] + " | " + board[1] + " | " + board[2])
    print("--+---+--")
    print(board[3] + " | " + board[4] + " | " + board[5])
    print("--+---+--")
    print(board[6] + " | " + board[7] + " | " + board[8])
    print()


def check_winner(player):
    winning_combinations = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_combinations:
        if board[a] == board[b] == board[c] == player:
            return True

    return False


def board_full():
    return " " not in board


def minimax(is_computer_turn):
    if check_winner(COMPUTER):
        return 1

    if check_winner(HUMAN):
        return -1

    if board_full():
        return 0

    if is_computer_turn:
        best_score = -float("inf")

        for i in range(9):
            if board[i] == " ":
                board[i] = COMPUTER
                score = minimax(False)
                board[i] = " "
                best_score = max(best_score, score)

        return best_score

    else:
        best_score = float("inf")

        for i in range(9):
            if board[i] == " ":
                board[i] = HUMAN
                score = minimax(True)
                board[i] = " "
                best_score = min(best_score, score)

        return best_score


def computer_move():
    best_score = -float("inf")
    best_move = None

    for i in range(9):
        if board[i] == " ":
            board[i] = COMPUTER
            score = minimax(False)
            board[i] = " "

            if score > best_score:
                best_score = score
                best_move = i

    board[best_move] = COMPUTER


# Main Game
print("Tic-Tac-Toe")
print("Positions are numbered 1 to 9:")
print("1 | 2 | 3")
print("--+---+--")
print("4 | 5 | 6")
print("--+---+--")
print("7 | 8 | 9")

while True:
    display_board()

    # Human move
    while True:
        try:
            move = int(input("Enter your move (1-9): ")) - 1

            if 0 <= move < 9 and board[move] == " ":
                board[move] = HUMAN
                break
            else:
                print("Invalid move. Try again.")

        except ValueError:
            print("Enter a number from 1 to 9.")

    if check_winner(HUMAN):
        display_board()
        print("Human Wins")
        break

    if board_full():
        display_board()
        print("Game Draw")
        break

    # Computer move
    computer_move()
    print("Computer's move:")

    if check_winner(COMPUTER):
        display_board()
        print("Computer Wins")
        break

    if board_full():
        display_board()
        print("Game Draw")
        break
```

```python
# Vacuum Cleaner Agent

# Step 1: Read input
vacuum_position = input("Enter vacuum position (L/R): ").upper()
left_status = input("Enter left room status (Dirty/Clean): ").capitalize()
right_status = input("Enter right room status (Dirty/Clean): ").capitalize()

print("\nInitial State:")
print("Vacuum Position:", vacuum_position)
print("Left Room:", left_status)
print("Right Room:", right_status)

# Step 2: Perform actions until both rooms are clean
while left_status == "Dirty" or right_status == "Dirty":

    if vacuum_position == "L":

        if left_status == "Dirty":
            print("\nAction: SUCK")
            left_status = "Clean"
            print("Left room is now Clean")

        else:
            print("\nAction: RIGHT")
            vacuum_position = "R"
            print("Vacuum moved to Right")

    elif vacuum_position == "R":

        if right_status == "Dirty":
            print("\nAction: SUCK")
            right_status = "Clean"
            print("Right room is now Clean")

        else:
            print("\nAction: LEFT")
            vacuum_position = "L"
            print("Vacuum moved to Left")

# Step 3: Goal reached
print("\nAction: STOP")
print("Goal reached!")
print("Both rooms are clean.")
```
